```python
import matplotlib.pyplot as plt
import numpy as np

data1 = '20261119_130456'

with open('./model/MyOriginal_train_loss_' + data1 + '.csv',encoding='utf-8')as f:
    line1_all = f.readlines()

#MyOriginal_val_loss_20260917_195918.csv
with open('./model/MyOriginal_val_loss_' + data1 + '.csv',encoding='utf-8')as f:
    line2_all = f.readlines()


i = 0
x = []
loss = []
crf_loss = []
ce_loss = []
length_loss = []
rouge1 = []
rouge2 = []
rougeL = []
bert_score = []
bleu_score = []
wer = []
return1 = []
v_i = 0
v_x = []
v_rouge1 = []
v_rouge2 = []
v_rougeL = []
v_bert = []
v_bleu = []
v_wer = []
v_return = []

#line2 = line2_all[0].replace( "\n", "")
#len_tr_loader = int(line2)

for i, line1 in enumerate( line1_all ):
    if i == 0:
        line1 = line1.replace( "\n", "")
        len_tr_loader = int(line1)
    else:
        line1_split = line1.split(",")
        #print( line1_split )
        i += 1
        #x.append( i / len_tr_loader )
        #x.append( float(line1_split[0].split(' ')[0]) - 0.27 )
        x.append( float(line1_split[0].split(' ')[0]) )
        loss.append( float(line1_split[1].split(' ')[1]) )
        crf_loss.append( float(line1_split[2].split(' ')[1]) )
        ce_loss.append( float(line1_split[3].split(' ')[1]) )
        length_loss.append( float(line1_split[4].split(' ')[1]) )
        rouge1.append( float(line1_split[5].split(' ')[1]) )
        rouge2.append( float(line1_split[6].split(' ')[1]) )
        rougeL.append( float(line1_split[7].split(' ')[1]) )
        bert_score.append( float(line1_split[8].split( ' ' )[1]))
        bleu_score.append( float(line1_split[9].split( ' ' )[1]))
        wer.append( float(line1_split[10].split( ' ' )[1]))
        return1.append( float(line1_split[11].split( ' ' )[1]))

#print( "pad:", pad )

for i, line2 in enumerate( line2_all ):
    if i == 0:
        line2 = line2.replace( "\n", "")
        len_val_loader = int(line2)
    else:
        #print( "line2:", line2 )
        line2_split = line2.split(",")
        v_i += 1
        #v_x.append( float(line2_split[0].split(' ')[0] ) )
        v_x.append( float(line2_split[0].split(' ')[0] ) )
        v_rouge1.append( float(line2_split[1].split(' ')[1]) )
        v_rouge2.append( float(line2_split[2].split(' ')[1]) )
        v_rougeL.append( float(line2_split[3].split(' ')[1]) )
        v_bert.append( float(line2_split[4].split(' ')[1]) )
        v_bleu.append( float(line2_split[5].split(' ')[1]) )
        v_wer.append( float(line2_split[6].split(' ')[1] ) )
        v_return.append( float(line2_split[7].split(' ')[1] ) )

#print( x )
#print( v_x )


#print( v_return )

plt.plot( x, loss, label="Train loss")
#plt.plot( v_x, v_loss, label="Val Loss" )
plt.title( "Loss")
plt.xlabel( 'steps')
plt.ylabel( 'Loss')
plt.legend()
#plt.ylim( 0, 7 )
plt.show()

#print( ctc )

plt.plot( x, crf_loss, label="Train crf loss")
#plt.plot( v_x, v_ctc, label="Val crf Loss" )
plt.title( "crf Loss")
plt.xlabel( 'steps')
plt.ylabel( 'crf Loss')
plt.legend()
#plt.ylim( 0, 5 )
plt.show()

plt.plot( x, ce_loss, label="Train ce loss")
#plt.plot( v_x, v_kd, label="Val ca Loss" )
plt.title( "ce Loss")
plt.xlabel( 'steps')
plt.ylabel( 'ce Loss')
plt.legend()
#plt.ylim( 0, 3 )
plt.show()

plt.plot( x, length_loss, label="Train length loss")
#plt.plot( v_x, v_kd, label="Val ca Loss" )
plt.title( "length Loss")
plt.xlabel( 'steps')
plt.ylabel( 'length Loss')
plt.legend()
#plt.ylim( 0, 3 )
plt.show()

plt.plot( x, rouge1, label = 'Train rouge1')
plt.plot( v_x, v_rouge1, label="Val rouge1")
plt.title( "rouge1")
plt.xlabel( 'steps')
plt.ylabel( 'rouge1')
plt.legend()
#plt.xlim( 9, 10 )
#plt.ylim( 0, 120)
plt.show()

plt.plot( x, rouge2, label = 'Train rouge2')
plt.plot( v_x, v_rouge2, label="Val rouge2")
plt.title( "rouge2")
plt.xlabel( 'steps')
plt.ylabel( 'rouge2')
plt.legend()
#plt.xlim( 9, 10 )
#plt.ylim( 0, 120)
plt.show()

plt.plot( x, rougeL, label = 'Train rougeL')
plt.plot( v_x, v_rougeL, label="Val rougeL")
plt.title( "rougeL")
plt.xlabel( 'steps')
plt.ylabel( 'rougeL')
plt.legend()
#plt.xlim( 9, 10 )
#plt.ylim( 0, 120)
plt.show()

plt.plot( x, bert_score, label = 'Train bert_score')
plt.plot( v_x, v_bert, label="Val bert_score")
plt.title( "bert_score")
plt.xlabel( 'steps')
plt.ylabel( 'bert_score')
plt.legend()
#plt.ylim( 0, 120 )
plt.show()


plt.plot( x, bleu_score, label = 'Train bleu_score')
plt.plot( v_x, v_bleu, label="Val bleu_score")
plt.title( "bleu_score")
plt.xlabel( 'Epochs')
plt.ylabel( 'bleu_score')
plt.legend()
#plt.ylim( 0, 120 )
plt.show()

plt.plot( x, wer, label = 'Train wer')
plt.plot( v_x, v_wer, label="Val wer")
plt.title( "wer")
plt.xlabel( 'Epochs')
plt.ylabel( 'wer')
plt.legend()
#plt.ylim( 0, 120 )
plt.show()

plt.plot( x, return1, label = 'Train return')
plt.plot( v_x, v_return, label="Val return")
plt.title( "return")
plt.xlabel( 'Epochs')
plt.ylabel( 'return')

plt.legend()
#plt.ylim( 0, 120 )
plt.show()
```
